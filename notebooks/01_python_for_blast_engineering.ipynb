# Python for Blast Engineering

Portable reconstruction of the first BlastDesign-AI learning notebook. This notebook is self-contained and can be restarted and run from top to bottom.

```python
from pathlib import Path
import sys
import math
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

PROJECT_ROOT = Path.cwd().resolve()
if PROJECT_ROOT.name == 'notebooks':
    PROJECT_ROOT = PROJECT_ROOT.parent
sys.path.insert(0, str(PROJECT_ROOT / 'src'))

from blastdesign import calculate_blast_metrics

print('Mining AI environment is ready.')
print('Project root:', PROJECT_ROOT)
```

## Blast-design input variables

```python
bench_height_m = 10.0
hole_diameter_mm = 102.0
burden_m = 3.0
spacing_m = 3.6
subdrilling_m = 0.8
stemming_m = 2.5
explosive_density_kg_m3 = 1100.0
number_of_holes = 40

print(f'Bench height: {bench_height_m} m')
print(f'Hole diameter: {hole_diameter_mm} mm')
print(f'Burden × spacing: {burden_m} × {spacing_m} m')
print(f'Number of holes: {number_of_holes}')
```

## Calculate geometry, explosive mass and powder factor

```python
blast_result = calculate_blast_metrics(
    bench_height_m=bench_height_m,
    hole_diameter_mm=hole_diameter_mm,
    burden_m=burden_m,
    spacing_m=spacing_m,
    subdrilling_m=subdrilling_m,
    stemming_m=stemming_m,
    explosive_density_kg_m3=explosive_density_kg_m3,
    number_of_holes=number_of_holes,
)

result_table = pd.DataFrame(
    blast_result.items(), columns=['Parameter', 'Value']
)
result_table['Value'] = result_table['Value'].round(4)
display(result_table)
```

```python
print(
    f"Explosive mass per hole: "
    f"{blast_result['explosive_mass_per_hole_kg']:.2f} kg"
)
print(
    f"Total explosive mass: "
    f"{blast_result['total_explosive_mass_kg']:.2f} kg"
)
print(
    f"Powder factor: "
    f"{blast_result['powder_factor_kg_m3']:.4f} kg/m³"
)

assert math.isclose(
    blast_result['explosive_mass_per_hole_kg'],
    74.60380915184177,
)
assert math.isclose(
    blast_result['total_explosive_mass_kg'],
    2984.152366073671,
)
print('First-notebook benchmark passed.')
```
